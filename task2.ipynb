## Intensity clipping

First we define some fixed thresholds $T_1$, $T_2$:

```python
T1 = 0.31
T2 = 0.59
```

### Using `ndarray.clip` method

**<a id="#">Task 2.3.1.</a>** Enter your solution below:

```python
img.clip(T1, T2)

plt.imshow(img, cmap='pink')
plt.colorbar()
```

### Not using the `ndarray.clip` method

**<a id="#">Task 2.3.2 (a).</a>** Enter your solution below:

```python
clip1 = img.copy()

rows = clip1.shape [0]
cols = clip1.shape [1]

for i in range (rows):
    for j in range (cols):
        clip1[i,j] = img [i,j]
        if clip1[i,j] < T1:
            clip1[i,j] = T1
        if clip1[i,j] > T2:
            clip1[i,j] = T2

plt.imshow(clip1, cmap='pink')
plt.colorbar()
```

**<a id="#">Task 2.3.2 (b).</a>** Enter your solution below:

```python
clip2 = img.copy()

for i in numpy.ndindex(img.shape):
    rows = i[0]
    cols = i[1]

    clip2[rows,cols] = img[rows,cols]

    if clip2[rows,cols] < T1:
        clip2[rows,cols] = T1
    if clip2[rows,cols] > T2:
        clip2 [rows,cols] = T2

plt.imshow(clip2, cmap='pink')
plt.colorbar()
```
