## Prewitt filters

**<a id="#">Task 4.1.1.</a>** Load the image `data/astronaut.png` and show it:

```python
img = plt.imread ('data/astronaut.png')
plt.imshow(img, 'gray')
```

**<a id="#">Task 4.1.2.</a>** Implement horizontal and vertical 3&times;3 Prewitt filters:

```python
def prewitt_h(img_in):
    result = numpy.zeros(img_in.shape)
    q = numpy.array ([[-1, 0, 1], [-1, 0, 1], [-1, 0, 1]])

    for i in range (1, img_in.shape[0] - 1):
        for j in range (1, img_in.shape[1] - 1):
            s = 0
            for u in range (-1, 2):
                for v in range (-1, 2):
                    s += img_in[i + u, j + v] * q[u + 1, v + 1]
                result[i, j] = s

    return result
    
def prewitt_v(img_in):
    result = numpy.zeros(img_in.shape)
    q = numpy.array ([[1, 1, 1], [0, 0, 0], [-1, -1, -1]])

    for i in range (1, img_in.shape[0] - 1):
        for j in range (1, img_in.shape[1] - 1):
            s = 0
            for u in range (-1, 2):
                for v in range (-1, 2):
                    s += img_in[i + u, j + v] * q[u + 1, v + 1]
                result[i, j] = s

    return result
```

**<a id="#">Task 4.1.3.</a>** Test your above solution using the functions `prewitt_h` and `prewitt_v` for the previously loaded image and show the result:

```python
gx = prewitt_h(img)
gy = prewitt_v(img)

plt.imshow(gx, cmap="gray")
plt.colorbar()

plt.imshow(gy, cmap="gray")
plt.colorbar()
```

**<a id="#">Task 4.1.4.</a>** Compare your results with the correct result image `data/astronaut_prewitt_h.tiff` and `data/astronaut_prewitt_v.tiff`:

```python
import skimage.io as ski

imgh = ski.imread("data/astronaut_prewitt_h.tiff").astype(float)
imgv = ski.imread("data/astronaut_prewitt_v.tiff").astype(float)

gx = prewitt_h(img)
gy = prewitt_v(img)

print(numpy.allclose(gx, imgh, atol=1/255))
print(numpy.allclose(gy, imgv, atol=1/255))
```

## Sobel filters (bonus)

**<a id="#">Task 4.3.1.</a>** Implement the two functions `sobel_h` and `sobel_v` for horizontal and vertical 3&times;3 Sobel filtering, respectively:

```python
def sobel_h(img_in):
    result = numpy.zeros(img_in.shape)
    q = numpy.array ([[-1, 0, 1], [-2, 0, 2], [-1, 0, 1]])

    for i in range (1, img_in.shape[0] - 1):
        for j in range (1, img_in.shape[1] - 1):
            s = 0
            for u in range (-1, 2):
                for v in range (-1, 2):
                    s += img_in[i + u, j + v] * q[u + 1, v + 1]
                result[i, j] = s

    return result
    
def sobel_v(img_in):
    result = numpy.zeros(img_in.shape)
    q = numpy.array ([[-1, -2, -1], [0, 0, 0], [1, 2, 1]])

    for i in range (1, img_in.shape[0] - 1):
        for j in range (1, img_in.shape[1] - 1):
            s = 0
            for u in range (-1, 2):
                for v in range (-1, 2):
                    s += img_in[i + u, j + v] * q[u + 1, v + 1]
                result[i, j] = s

    return result
```

**<a id="#">Task 4.3.2.</a>** Test your above solution using the functions `sobel_h` and `sobel_v` for the previously loaded image and show the result:

```python
gx2 = sobel_h(img)
gy2 = sobel_v(img)

plt.imshow(gx2, cmap="gray")
plt.colorbar()

plt.imshow(gy2, cmap="gray")
plt.colorbar()
```
