## Linear filtering by convolution (mean filter)

**<a id="#">Task 3.1.1.</a>** Load the image `data/astronaut.png` and show it:

```python
img = plt.imread ('data/astronaut.png')
plt.imshow(img, 'gray')
```

**<a id="#">Task 3.1.2.</a>** Implement a mean filter:

```python
def meanfilter(img_in, size):
    result = numpy.zeros(img_in.shape)
    n = (size - 1) // 2
    for p in numpy.ndindex(img_in.shape):
        neighborhood_sum = 0
        count = 0

        for i in range (p[0]-n, p[0]+n+1):
            for j in range (p[1]-n,p[1]+n+1):

                if 0 <= i < img_in.shape[0] and 0 <= j < img_in.shape[1]:
                     neighborhood_sum += img_in[i, j]
                     count += 1
        
        result[p[0], p[1]] = neighborhood_sum / count
            
    return result
```

**<a id="#">Task 3.1.3.</a>** Test your implementation from above by using the function `meanfilter` for the previously loaded image and show the result:

```python
img2 = meanfilter (img, size=5)

plt.imshow(img2, cmap='gray')
plt.colorbar()

plt.show()
```
